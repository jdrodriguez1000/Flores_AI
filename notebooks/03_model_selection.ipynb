## Objetivo

Evaluar y seleccionar el algoritmo de clasificaciÃ³n Ã³ptimo para el clasificador multi-clase de especies Iris (setosa, versicolor, virginica), aplicando el principio de **Simplicidad Primero**: dado rendimiento equivalente, se prefiere el modelo de menor complejidad computacional y menor nÃºmero de hiperparÃ¡metros.

## Trazabilidad

| Artefacto | Referencia |
|:----------|:-----------|
| KPI de negocio | BRD Â§4.1 â€” KPI-01: F1-Score Macro >= 0.95 (objetivo), Accuracy Test >= 95% |
| Contrato de interfaz | SpecDD Â§9 â€” `train_and_evaluate()` |
| Datos de entrada | Feature Store Gold: `data/gold/X_gold.csv`, `data/gold/y_gold.csv` |
| MÃ³dulo de training | `src/training/trainer.py` |

## KPIs de AceptaciÃ³n (BRD Â§4.1)

| KPI | Umbral MÃ­nimo (RED) | Umbral Objetivo (GREEN) |
|:----|:--------------------|:------------------------|
| Accuracy Test | >= 0.92 | >= 0.95 |
| F1-Score Macro | >= 0.92 | >= 0.95 |
| F1-Score por Clase | >= 0.90 c/u | >= 0.95 c/u |

**Baseline de referencia:** RandomForest sin hiperparametrizar â†’ 95.20% Â± 3.53% accuracy (CV 5-fold).

```python
# ---------------------------------------------------------------------------
# Imports y resoluciÃ³n de path del proyecto
# ---------------------------------------------------------------------------
import sys
import os

# Asegurar que la raÃ­z del proyecto estÃ© en el path (notebook corre desde notebooks/)
sys.path.insert(0, os.path.join(os.path.dirname(os.path.abspath('__file__')), ".."))

import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from src.training.trainer import train_and_evaluate
from src import config

print(f"Python path configurado. PROJECT_ROOT: {config.PROJECT_ROOT}")
print(f"Constantes activas â€” RANDOM_STATE={config.RANDOM_STATE}, TEST_SIZE={config.TEST_SIZE}, CV_FOLDS={config.CV_FOLDS}")
```

```python
# ---------------------------------------------------------------------------
# Carga de datos Gold (Feature Store)
# Fuente: data/gold/X_gold.csv (147, 4) y data/gold/y_gold.csv (147, 1)
# NO aplicar StandardScaler aquÃ­ â€” lo gestiona build_pipeline() dentro de train_and_evaluate()
# ---------------------------------------------------------------------------
X = pd.read_csv("../data/gold/X_gold.csv").to_numpy()           # shape (147, 4), dtype float64
y = pd.read_csv("../data/gold/y_gold.csv")["species"].to_numpy() # shape (147,), dtype object

print(f"X shape: {X.shape} | dtype: {X.dtype}")
print(f"y shape: {y.shape} | clases Ãºnicas: {np.unique(y)}")
print(f"DistribuciÃ³n de clases: {dict(zip(*np.unique(y, return_counts=True)))}")
```

```python
# ---------------------------------------------------------------------------
# DefiniciÃ³n de candidatos
# Criterio de selecciÃ³n del espacio: Simplicidad Primero
#   - LogisticRegression: modelo lineal, mÃ­nima complejidad, alta interpretabilidad
#   - RandomForestClassifier: ensemble no-lineal, baseline de referencia del proyecto
#   - SVC (probability=True): mÃ¡quina de vectores de soporte, kernel RBF, no-lineal
# Todos usan hiperparÃ¡metros por defecto (vanilla) para evaluaciÃ³n justa
# ---------------------------------------------------------------------------
candidates: dict = {
    "LogisticRegression": LogisticRegression(
        random_state=config.RANDOM_STATE,
        max_iter=1000,    # aumentado para garantizar convergencia con StandardScaler aplicado
    ),
    "RandomForestClassifier": RandomForestClassifier(
        random_state=config.RANDOM_STATE,
    ),
    "SVC": SVC(
        probability=True,  # requerido por SpecDD Â§9 para salida probabilÃ­stica
        random_state=config.RANDOM_STATE,
    ),
}

print("Candidatos registrados:")
for name, clf in candidates.items():
    print(f"  - {name}: {clf.__class__.__module__}.{clf.__class__.__name__}")
```

```python
# ---------------------------------------------------------------------------
# Loop de evaluaciÃ³n
# Cada candidato usa train_and_evaluate() de src/training/trainer.py
# NO se reimplementa la lÃ³gica de training â€” trazable a SpecDD Â§9
# ---------------------------------------------------------------------------
results: list[dict] = []
fitted_pipelines: dict = {}

for name, classifier in candidates.items():
    pipeline, metrics = train_and_evaluate(
        X=X,
        y=y,
        classifier=classifier,
        test_size=config.TEST_SIZE,
        random_state=config.RANDOM_STATE,
        cv_folds=config.CV_FOLDS,
    )
    fitted_pipelines[name] = pipeline

    row = {
        "Algoritmo": name,
        "accuracy_cv_mean": round(metrics["accuracy_cv_mean"], 4),
        "accuracy_cv_std":  round(metrics["accuracy_cv_std"],  4),
        "accuracy_test":    round(metrics["accuracy_test"],    4),
        "f1_macro_test":    round(metrics["f1_macro_test"],    4),
        "f1_setosa":        round(metrics["f1_per_class"]["setosa"],     4),
        "f1_versicolor":    round(metrics["f1_per_class"]["versicolor"], 4),
        "f1_virginica":     round(metrics["f1_per_class"]["virginica"],  4),
        "n_train":          metrics["n_train"],
        "n_test":           metrics["n_test"],
    }
    results.append(row)
    print(f"[OK] {name} evaluado â€” F1 Macro: {row['f1_macro_test']:.4f} | Accuracy Test: {row['accuracy_test']:.4f} | CV Mean: {row['accuracy_cv_mean']:.4f} +/- {row['accuracy_cv_std']:.4f}")

# Tabla comparativa
df_results = pd.DataFrame(results).set_index("Algoritmo")
print("\n=== TABLA COMPARATIVA ===")
print(df_results.to_string())
```

## AnÃ¡lisis de Resultados por Candidato

### LogisticRegression

Modelo lineal de referencia. Se aplica sobre datos estandarizados (StandardScaler dentro del Pipeline). La correlaciÃ³n alta entre `petal_length`/`petal_width` (r=0.962) genera multicolinealidad (VIF elevado), pero con la regularizaciÃ³n L2 por defecto (`C=1.0`) el modelo la gestiona sin necesidad de PCA o eliminaciÃ³n de variables. Es el modelo de **menor complejidad** del conjunto: sin hiperparÃ¡metros estructurales, entrenamiento O(n*p), inferencia O(p).

**Zona critica (solapamiento versicolor/virginica):** Al ser un separador lineal, tiene el mayor riesgo de error en la zona `petal_width` [1.4â€“1.8 cm] donde ~17 instancias se solapan. Los F1 de las clases versicolor y virginica son el indicador directo de este riesgo.

### RandomForestClassifier

Ensemble de 100 Ã¡rboles de decisiÃ³n (por defecto). No-lineal, robusto a la multicolinealidad por naturaleza (feature sampling por Ã¡rbol). Es el **baseline de referencia** del proyecto (95.20% Â± 3.53% en CV). Mayor complejidad computacional que LogÃ­stica: entrenamiento O(n*p*log(n)*T), inferencia O(T*profundidad). No requiere escalado pero el Pipeline lo incluye sin impacto negativo.

### SVC (kernel RBF)

MÃ¡quina de vectores de soporte con kernel gaussiano. Puede capturar separaciones no-lineales en el espacio de caracterÃ­sticas proyectado. Requiere `probability=True` para salida probabilÃ­stica (Platt calibration). Alta sensibilidad a la escala â€” favorecido por el StandardScaler del Pipeline. Complejidad de entrenamiento O(nÂ²) a O(nÂ³), lo que lo hace el mÃ¡s costoso del conjunto para producciÃ³n.

## JustificaciÃ³n de la SelecciÃ³n del Algoritmo Ganador

### Criterio de SelecciÃ³n Aplicado

**Principio de Simplicidad Primero** (CLAUDE.md y BRD Â§4.1):

> Si `LogisticRegression` alcanza `f1_macro_test >= 0.95`, debe ser seleccionado sobre RandomForest o SVC aunque estos tengan mÃ©tricas marginalmente superiores.

El criterio opera en dos pasos:
1. Filtro de umbral: el candidato debe cumplir `f1_macro_test >= 0.95` (GREEN)
2. Desempate por complejidad: entre los que superan el umbral, se selecciona el de menor complejidad computacional y menor nÃºmero de hiperparÃ¡metros

### Matriz de Complejidad Relativa

| Algoritmo | HiperparÃ¡metros libres | Complejidad entrenamiento | Complejidad inferencia | Interpretabilidad |
|:----------|:----------------------|:--------------------------|:-----------------------|:------------------|
| LogisticRegression | 2 (C, solver) | O(nÂ·pÂ·iter) | O(p) | Alta (coeficientes lineales) |
| RandomForestClassifier | 5+ (n_estimators, max_depth, ...) | O(nÂ·pÂ·log(n)Â·T) | O(TÂ·depth) | Media (feature importance) |
| SVC | 3 (C, gamma, kernel) | O(nÂ²) a O(nÂ³) | O(n_svÂ·p) | Baja (caja negra) |

```python
# ---------------------------------------------------------------------------
# AplicaciÃ³n explÃ­cita del criterio de selecciÃ³n
# ---------------------------------------------------------------------------
F1_THRESHOLD = 0.95  # Umbral GREEN (BRD Â§4.1 - KPI-01)

# Orden de complejidad ascendente (Simplicidad Primero)
# Justificacion del orden:
#   1. LogisticRegression: O(n*p) entrenamiento, O(p) inferencia â€” minima complejidad
#   2. RandomForestClassifier: O(n*p*log(n)*T) entrenamiento, O(T*depth) inferencia â€” medio, paralelizable
#   3. SVC: O(n^2) a O(n^3) entrenamiento, O(n_sv*p) inferencia + Platt calibration â€” mayor costo en produccion
COMPLEXITY_ORDER = ["LogisticRegression", "RandomForestClassifier", "SVC"]

print(f"Umbral de aceptacion F1 Macro (GREEN): {F1_THRESHOLD}")
print()

# Filtrar candidatos que superan el umbral
qualified = [
    row for row in results
    if row["f1_macro_test"] >= F1_THRESHOLD
]

print("Candidatos que superan el umbral GREEN (f1_macro_test >= 0.95):")
for row in qualified:
    print(f"  - {row['Algoritmo']}: f1_macro_test={row['f1_macro_test']:.4f}")

print()

# Seleccionar el mÃ¡s simple entre los calificados
if qualified:
    qualified_sorted = sorted(
        qualified,
        key=lambda r: COMPLEXITY_ORDER.index(r["Algoritmo"])
        if r["Algoritmo"] in COMPLEXITY_ORDER else len(COMPLEXITY_ORDER)
    )
    winner = qualified_sorted[0]
    selection_reason = "Simplicidad Primero: modelo mÃ¡s simple entre los que superan f1_macro_test >= 0.95"
else:
    # Si ninguno supera el umbral, se selecciona el de mayor f1_macro_test
    winner = max(results, key=lambda r: r["f1_macro_test"])
    selection_reason = "NingÃºn candidato supera el umbral GREEN â€” se selecciona el de mayor f1_macro_test (requiere revisiÃ³n)"

print(f"ALGORITMO SELECCIONADO: {winner['Algoritmo']}")
print(f"RazÃ³n: {selection_reason}")
print()
print("MÃ©tricas del ganador:")
print(f"  accuracy_cv_mean : {winner['accuracy_cv_mean']:.4f}")
print(f"  accuracy_cv_std  : {winner['accuracy_cv_std']:.4f}")
print(f"  accuracy_test    : {winner['accuracy_test']:.4f}")
print(f"  f1_macro_test    : {winner['f1_macro_test']:.4f}")
print(f"  f1_setosa        : {winner['f1_setosa']:.4f}")
print(f"  f1_versicolor    : {winner['f1_versicolor']:.4f}")
print(f"  f1_virginica     : {winner['f1_virginica']:.4f}")

# VerificaciÃ³n de KPIs BRD Â§4.1
print()
print("Verificacion de KPIs BRD Â§4.1:")
kpi_accuracy = winner['accuracy_test'] >= 0.95
kpi_f1_macro = winner['f1_macro_test'] >= 0.95
kpi_f1_per_class = all([
    winner['f1_setosa']     >= 0.95,
    winner['f1_versicolor'] >= 0.95,
    winner['f1_virginica']  >= 0.95,
])
print(f"  [{'PASS' if kpi_accuracy else 'FAIL'}] Accuracy Test >= 0.95: {winner['accuracy_test']:.4f}")
print(f"  [{'PASS' if kpi_f1_macro else 'FAIL'}] F1 Macro >= 0.95: {winner['f1_macro_test']:.4f}")
print(f"  [{'PASS' if kpi_f1_per_class else 'REVIEW'}] F1 por clase >= 0.95: setosa={winner['f1_setosa']:.4f} versicolor={winner['f1_versicolor']:.4f} virginica={winner['f1_virginica']:.4f}")

all_kpis_green = kpi_accuracy and kpi_f1_macro and kpi_f1_per_class
print()
print(f"Estado final: {'[GREEN] Todos los KPIs objetivo superados' if all_kpis_green else '[YELLOW] Revisar KPIs individuales â€” puede requerir optimizacion de hiperparametros (F3-T06)'}")
```

## Conclusion

### Algoritmo Seleccionado: ver celda anterior

El proceso de selecciÃ³n aplica el **Principio de Simplicidad Primero** en dos etapas:

1. **Filtro de umbral GREEN:** Se retienen Ãºnicamente los candidatos con `f1_macro_test >= 0.95` (KPI-01, BRD Â§4.1).
2. **Desempate por complejidad:** Entre los calificados, se selecciona `LogisticRegression` si supera el umbral, dado que es el modelo de menor complejidad computacional (O(nÂ·p) inferencia vs. O(TÂ·depth) para RandomForest), menor nÃºmero de hiperparÃ¡metros estructurales, y mayor interpretabilidad directa (coeficientes lineales exportables para XAI).

### JustificaciÃ³n tÃ©cnica adicional

- La alta separabilidad de la clase `setosa` (hiperplano natural en `petal_length`/`petal_width`) favorece a los modelos lineales.
- La multicolinealidad `petal_length`/`petal_width` (r=0.962) no es un bloqueador para LogisticRegression con regularizaciÃ³n L2 activa.
- La zona de solapamiento versicolor/virginica (~17 instancias en `petal_width` [1.4â€“1.8 cm]) representa el lÃ­mite fÃ­sico del dataset â€” cualquier modelo que supere el 95% F1 Macro en esta distribuciÃ³n es Ã³ptimo para el dominio.
